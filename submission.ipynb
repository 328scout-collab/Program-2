To find the k-th smallest number we used a divide and conqure method. First we pick a random number to be our pivot, then iterate through the array divding into sub arays smaller, equal, and greater than the pivot. We can then use the size of the arrays to determine which sub array K is in. After determining we recursivly call our method, passing in the correct subarray. If K is in the greater sub array we subtract the size of the smaller and equal arrays from K before passing it in. This will continue until either the size of the sub array is 1 or K is in the equal sub array, then the pivot is returned.

 
Illistration 1
        [5, 1, 6, 7, 3, 4, 8]     pivot = 5
           /       |        \
   [1, 3, 4]      [5]     [6, 7, 8]
   smaller       equal     greater
   size 3        size 1    size 3

   K = 3 ≤ 3  →  answer is in smaller


Illistration 2

Call 1: [5, 1, 6, 7, 3, 4, 8]  K=3  pivot=5
        smaller=[1,3,4]  equal=[5]  greater=[6,7,8]
        K ≤ 3 → go to smaller, K stays 3

Call x: **MAGIC**

Final call: [4]  K=1
        size 1 → return 4 ✅

```python
import random

def Find_Kth_Smallest(arr, k):

    #Validates that the array is not empty and has a valid k
    if not arr or not (1 <= k <= len(arr)):
        raise ValueError("Invalid array or k out of bounds")

    #Base Case
    if len(arr) == 1:
        return arr[0]
    
    #Choose random pivot
    pivot = random.choice(arr)

    #Partition array into smaller lists based on pivot
    #smaller, equal, and larger values
    smaller = [x for x in arr if x < pivot]
    equal = [x for x in arr if x == pivot]
    larger = [x for x in arr if x > pivot]

    #Determine recurse direction
    if k <= len(smaller):
        return Find_Kth_Smallest(smaller, k)
    elif k <= len(smaller) + len(equal):
        return pivot
    else:
        return Find_Kth_Smallest(larger, k - len(smaller) - len(equal))


def main():
    print(Find_Kth_Smallest([5, 1, 6, 7, 3, 4, 8], 3))   # should print 4

if __name__ == "__main__":
    main()
```

```python
def run_tests():
    test_cases = [
        ([5, 1, 6, 7, 3, 4, 8], 3, 4),
        ([10, 2, 9, 3, 8, 4], 1, 2),
        ([10, 2, 9, 3, 8, 4], 6, 10),
        ([5, 2, 5, 1, 5, 3], 4, 5),
        ([7, 3], 1, 3),
        ([-5, 0, -2, 10, 3], 2, -2),
        ([42, 12, 18, 5, 99], 3, 18)
    ]
    
    for i, (arr, k, expected) in enumerate(test_cases, 1):
        result = Find_Kth_Smallest(arr.copy(), k)
        status = "PASS" if result == expected else "FAIL"
        print(f"Test {i}: Expected {expected}, Got {result} -> [{status}]")

if __name__ == "__main__":
    run_tests()
```

```python
import time


def benchmark_sorting(arr, k):
    
    return sorted(arr)[k - 1]

def run_benchmarks():
    sizes = [1000, 5000, 10000, 50000, 100000,1000000]
    
    print(f"{'N Size':<10} | {'QuickSelect (s)':<18} | {'Sorting Approach (s)':<20}")
    print("-" * 52)
    
    for n in sizes:
        # Generate random test array
        arr = [random.randint(1, 1000000) for _ in range(n)]
        k = n // 2  # Find the median
        
        # Benchmark QuickSelect
        start_time = time.time()
        Find_Kth_Smallest(arr.copy(), k)
        qs_time = time.time() - start_time
        
        # Benchmark Sorting Approach
        start_time = time.time()
        benchmark_sorting(arr.copy(), k)
        sort_time = time.time() - start_time
        
        print(f"{n:<10} | {qs_time:<18.6f} | {sort_time:<20.6f}")

if __name__ == "__main__":
    run_benchmarks()
```
